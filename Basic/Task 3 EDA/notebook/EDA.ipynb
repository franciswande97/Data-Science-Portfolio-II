Explore the relationships among the remaining features and identify any patterns that warrant further investigation.

```python
remaining_features = [
    "CRIM", "ZN", "INDUS", "CHAS", "NOX",
    "AGE", "DIS", "RAD", "PTRATIO", "B"
]

fig, axes = plt.subplots(5, 2, figsize=(14, 22))

for ax, feature in zip(axes.flatten(), remaining_features):
    ax.scatter(df[feature], df["MEDV"])
    ax.set_xlabel(feature)
    ax.set_ylabel("MEDV")
    ax.set_title(f"{feature} vs MEDV")

plt.tight_layout()
plt.savefig("../outputs/remaining_features_vs_medv.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

### Scatter Plot Relationship Analysis

Scatter plots were generated for all explanatory variables against the target variable (MEDV) to avoid selectively focusing only on variables that appeared important.

Most variables did not show an obvious relationship with MEDV from visual inspection. Variables such as PTRATIO, B, DIS, ZN, INDUS, AGE, CRIM, and TAX showed substantial dispersion, repeated values, or no clear directional pattern.

RM showed a strong positive relationship with MEDV. The observations are cluster tightly together and there is a general tendency for MEDV to increase as RM increases. The relationship appears approximately linear.

LSTAT showed the clearest negative relationship with MEDV. As LSTAT increases, MEDV generally decreases. However, the relationship is not perfectly linear and shows some curvature.

Based on the visual analysis, RM and LSTAT appear to have the most notable relationships with the target variable, while the remaining features do not show an obvious relationship from their scatter plots alone. These observations will be validated quantitatively using the correlation matrix.

```python
#Correlation matrix to understand the relationships between different features in the dataset
correlation_matrix = df.corr(numeric_only=True)

correlation_matrix
```

```python
#Visualizing the correlation matrix using a heatmap to easily identify strong correlations between features
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.savefig("../outputs/correlation_heatmap.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
correlation_matrix["MEDV"].sort_values(ascending=False)
```

Areas with larger homes tend to command higher property values, while neighborhoods with higher socioeconomic disadvantage tend to have lower property values. Environmental and infrastructure-related factors such as crime, taxation, pollution, and industrial development also appear to influence housing prices, although their relationships are weaker than RM and LSTAT.

## Key Findings

- **RM (average number of rooms)** showed a positive relationship with MEDV. The scatter plot showed a general upward trend, and the observations were tightly cluster together. The correlation coefficient of **0.70** confirms a strong positive linear association.

- **LSTAT** showed the strongest negative relationship with MEDV. House values generally decreased as LSTAT increased. The correlation coefficient of **-0.74** confirms a strong negative association, although the scatter plot indicates that the relationship is not perfectly linear.

- **PTRATIO, INDUS, TAX, CRIM, and NOX** showed moderate negative correlations with MEDV, with correlation coefficients ranging from approximately **-0.43 to -0.51**. However, their scatter plots did not show clear or simple linear patterns, demonstrating why correlation should be interpreted alongside visual analysis.

- **ZN, B, DIS, CHAS, and AGE** showed weaker correlations with MEDV compared with RM and LSTAT. Their scatter plots also contained repeated values, concentrations, or substantial dispersion without an obvious directional pattern.

- **Repeated observations were present in several variables**, producing vertical or horizontal concentrations in some scatter plots. These observations were not automatically treated as outliers because repeated values can represent legitimate characteristics of the dataset.

- Overall, **RM and LSTAT were the two most prominent features in relation to MEDV**, while the remaining variables showed weaker or less visually obvious relationships.

## Business Implications

The analysis suggests that both property characteristics and socioeconomic conditions are associated with housing value.

- **Property valuation:** The positive relationship between RM and MEDV suggests that the number of rooms can be an important consideration when assessing property value.

- **Location and socioeconomic conditions:** The strong negative relationship between LSTAT and MEDV suggests that socioeconomic characteristics of an area may be important when evaluating housing markets.

- **Investment decisions:** Property investors and real-estate businesses could consider both property-level characteristics and neighborhood conditions when evaluating potential investments.

- **Pricing strategy:** Understanding the variables associated with higher or lower property values can support more informed pricing and valuation decisions.

- **Predictive modeling:** RM, LSTAT, and other variables with meaningful associations could be investigated further as potential predictors in a house-price prediction model.

## Limitations

- Correlation measures association rather than causation. The observed relationships do not establish that a feature directly causes changes in house prices.

- Correlation coefficients primarily measure linear relationships and may not fully represent non-linear patterns present in the data.

- Several variables contained repeated values and concentrated observations, which can influence the appearance of scatter plots.

- The analysis was exploratory and did not test statistical significance or the combined predictive power of the variables.

- The dataset represents a specific housing dataset and may not generalize to all housing markets or geographic regions.

- Further statistical analysis and predictive modeling are required to determine which variables provide the most reliable predictions of house value.

## Conclusion

The exploratory data analysis examined the structure, distributions, and relationships within the housing dataset to identify factors associated with median house value (MEDV).

The analysis found that **RM** and **LSTAT** had the most prominent relationships with MEDV. RM showed a Strong positive association, with higher numbers of rooms generally corresponding to higher house values, while LSTAT showed a strong negative association, with house values generally decreasing as LSTAT increased. Other variables, including PTRATIO, INDUS, TAX, CRIM, and NOX, showed moderate negative correlations but did not display equally clear visual relationships.

The analysis also identified repeated observations and concentrated values across several features. These were retained where they represented legitimate observations rather than automatically being classified as errors or outliers.

Overall, the EDA provides useful evidence for selecting potential features for future predictive modeling. However, the relationships identified represent associations rather than causation. A regression model can therefore be used in the next stage to evaluate how effectively these features can collectively predict house prices.
