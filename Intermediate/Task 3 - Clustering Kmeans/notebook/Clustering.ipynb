```python
import pandas as pd

churn_20 = pd.read_csv("../data/churn-bigml-20.csv")
churn_80 = pd.read_csv("../data/churn-bigml-80.csv")

print("20 dataset shape:", churn_20.shape)
print("80 dataset shape:", churn_80.shape)
```

```python
churn_df = pd.concat(
    [churn_20, churn_80],
    ignore_index=True
)
```

We'll exclude:

State - categorical location label,
Area code - identifier-like geographic code,
Churn - our known outcome; we will use it after clustering to interpret the segments

```python
X_cluster = churn_df.drop(
    columns=["State", "Area code", "Churn"]
).copy()

print("Clustering features:")
print(X_cluster.columns.tolist())
```

```python
X_cluster["International plan"] = (
    X_cluster["International plan"] == "Yes"
).astype(int)

X_cluster["Voice mail plan"] = (
    X_cluster["Voice mail plan"] == "Yes"
).astype(int)
```

```python
print(X_cluster.dtypes)
print(X_cluster.head())
```

```python
print("Missing values:")
print(X_cluster.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler_cluster = StandardScaler()

X_cluster_scaled = scaler_cluster.fit_transform(X_cluster)
```

```python
# Import necessary libraries
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
inertia = []
silhouette_scores = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_cluster_scaled)
    
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_cluster_scaled, labels)
    )
```

```python
#numerical results of elbow method and silhouette scores
for k, i, s in zip(K_range, inertia, silhouette_scores):
    print(f"K={k}: Inertia={i:.2f}, Silhouette Score={s:.3f}")
```

```python
#Visualize the elbow method results
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(K_range, inertia, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")
plt.xticks(list(K_range))

plt.show()
```

```python
#Silhouette score visualization
plt.figure(figsize=(8, 5))

plt.plot(K_range, silhouette_scores, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score by Number of Clusters")
plt.xticks(list(K_range))

plt.show()
```

K=2 was selected because it produced the highest silhouette score among the tested values and provided a reasonable balance between cluster compactness and model simplicity. However, the relatively low silhouette score indicates substantial overlap between the resulting customer segments.

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_cluster_scaled)
```

```python
churn_df["Cluster"] = cluster_labels
```

```python
print(churn_df["Cluster"].value_counts().sort_index())
```

```python
cluster_distribution = (
    churn_df["Cluster"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)

print(cluster_distribution)
```

```python
cluster_churn = pd.crosstab(
    churn_df["Cluster"],
    churn_df["Churn"],
    normalize="index"
) * 100

print(cluster_churn)
```

```python
#Average customer characteristics within each cluster
cluster_profiles = (
    X_cluster.assign(Cluster=cluster_labels)
    .groupby("Cluster")
    .mean()
)

cluster_profiles
```

```python
cluster_profiles.T
```

```python
#Calculating the difference between the cluster profiles
cluster_difference = (
    cluster_profiles.loc[1] - cluster_profiles.loc[0]
).sort_values(key=abs, ascending=False)

cluster_difference
```

```python
#Appying PCA to reduce all features to 2 dimensions for visualization
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_cluster_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=cluster_labels,
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Customer Segments Using K-Means Clustering")

plt.show()    
```

```python
churn_rates = churn_df.groupby("Cluster")["Churn"].mean() * 100

print("Churn rate by cluster:")
print(churn_rates)

difference = churn_rates[1] - churn_rates[0]

print(f"\nDifference in churn rate: {difference:.2f} percentage points")
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    ["Cluster 0", "Cluster 1"],
    churn_rates.values
)

plt.ylabel("Churn Rate (%)")
plt.xlabel("Customer Cluster")
plt.title("Churn Rate by Customer Cluster")

plt.show()
```

## Key Findings

- The combined dataset contained 3,333 customer records with no duplicate rows.
- K-Means clustering was evaluated for K values from 2 to 10.
- K=2 was selected because it produced the highest Silhouette Score (0.143), although the relatively low score indicates substantial overlap between the customer segments.
- Cluster 0 consisted primarily of customers with voicemail plans, while Cluster 1 consisted of customers without voicemail plans.
- The two clusters had broadly similar calling, international usage, and account-length characteristics.
- Cluster 0 had an observed churn rate of 8.68%, while Cluster 1 had an observed churn rate of 16.72%, a difference of 8.04 percentage points.
- PCA showed substantial overlap between the clusters, with the first two principal components explaining 24.04% of the total variance.

## Business Implications

The clustering analysis identified a customer segment with a substantially higher observed churn rate. Cluster 1 had a churn rate of 16.72%, compared with 8.68% for Cluster 0, suggesting that this segment could warrant additional retention analysis.

The primary characteristic separating the two clusters was voicemail-plan usage. Cluster 0 customers had a voicemail plan and averaged approximately 29 voicemail messages, while Cluster 1 customers did not have a voicemail plan. This provides a potential segmentation variable that could be investigated alongside other customer characteristics when developing retention strategies.

However, the clustering results should be treated as a starting point for customer segmentation rather than a definitive explanation of churn. The analysis identifies an association between the segment and observed churn rates, but does not establish that voicemail usage causes or prevents customer churn.

## Limitations

- The Silhouette Score of 0.143 indicates weak separation between the clusters, meaning that the customer segments overlap considerably.
- The PCA visualization showed substantial overlap, with the first two principal components explaining only 24.04% of the total variance.
- The clustering results were primarily differentiated by voicemail-plan usage, meaning the identified segments were not strongly distinct across the broader set of customer characteristics.
- K-Means requires the number of clusters to be specified in advance. Although the Elbow Method and Silhouette Score were used to select K=2, the Elbow Method did not produce a highly pronounced elbow.
- Churn was excluded from the clustering process, as it represents an existing outcome. It was only used afterward to interpret the resulting clusters.
- The observed difference in churn rates represents an association and should not be interpreted as evidence of causation.
